## 1. 📊 Estadística Descriptiva <a name="1"></a>

- Medidas de tendencia central (media, mediana, moda)  
- Medidas de dispersión (varianza, desviación estándar, rango intercuartílico)  
- Visualización de datos: histogramas, boxplots, distribuciones empíricas  

---

## 2. 🎲 Probabilidad y Distribuciones <a name="2"></a>

- Conceptos básicos de probabilidad (espacio muestral, eventos)  
- Variables aleatorias discretas y continuas  
- Distribuciones comunes en ciencia de datos: Bernoulli, Binomial, Poisson, Normal  
- Teorema de Bayes y probabilidad condicional  

---

## 3. 🔍 Inferencia Estadística <a name="3"></a>

- Estimaciones puntuales vs. por intervalo  
- Intervalos de confianza  
- Pruebas de hipótesis (hipótesis nula y alternativa, p‑valor)  
- Ejemplos con test t y χ²  

---

## 4. 📐 Modelado Estadístico Básico <a name="4"></a>

- Regresión lineal: estimación por mínimos cuadrados  
- Regresión logística: clasificación binaria  
- Interpretación de coeficientes y bondad de ajuste  
- Validación cruzada y evaluación de desempeño

---

## ✅ Resumen

- La estadística descriptiva te ayuda a **entender y explorar** datos.  
- La probabilidad y las distribuciones modelan incertidumbre y variabilidad.  
- La inferencia permite **generalizar** conclusiones de una muestra a una población.  
- El modelado estadístico conecta variables para hacer **predicciones y clasificación**.  
- La simulación, como el bootstrap, es una forma **experimental y práctica** de hacer inferencia.
