# Calculus for Data Science

## 🧭 Introduction

Calculus is the mathematical study of change. In data science, it provides the foundation for understanding concepts like optimization, model training, and continuous change in functions and distributions.

This notebook provides a practical and concise overview of the most relevant calculus concepts for data science applications. Rather than focusing on formal proofs or symbolic techniques, the focus here is on **intuition and use cases**.

We will explore:

- How derivatives help us understand change and optimize models
- How integrals relate to probability and continuous variables
- Real-world applications in machine learning and statistics
